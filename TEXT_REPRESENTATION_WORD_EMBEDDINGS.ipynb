```python
# @title Bag of Word `DATASET CREATION`

import numpy as np
import pandas as pd

df = pd.DataFrame({
    'text': [
        'people watch campusx',
        'campusx watch campusx',
        'people write comment',
        'campusx write comment'
    ],
    'output': [1,1,0,0]
})

df
```

```python
# @title USING `sklearn` `counVercorizer` perform `BAG OF WORDS`
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
```

```python
# @title BOW IMPLEMENTATTION
bow = cv.fit_transform(df['text'])
```

```python
# VOCABULARY
# It will assign index numver to all the unique words
print(cv.vocabulary_)
```

```python
# BOW ARRAY
bow.toarray()

#Check in order, 1.campusx  2.Comment 3. people 4. watch 5. write in sentence wise
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
```

```python
# @title Add New Sentence & Perform `BOW`
cv.transform(['campusx watch and write comment of campusx']).toarray()
```

```python
# @title Store Output of `BOW`

X = bow.toarray()
y = df['output']
```

```python
# @title N-Grams

import numpy as np
import pandas as pd

df = pd.DataFrame({
    'text': [
        'people watch campusx',
        'campusx watch campusx',
        'people write comment',
        'campusx write comment'
    ],
    'output': [1,1,0,0]
})

df
```

```python
# @title USING `N-GRAMS`
# BI grams

from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(ngram_range=(2,2))
# TI grams
# cv = CountVectorizer(ngram_range=(3,3))
# BI & TI grams
# cv = CountVectorizer(ngram_range=(1,3))
```

```python
bow_ngram = cv.fit_transform(df['text'])
```

```python
print(cv.vocabulary_)
```

```python
print(bow_ngram[0].toarray())
#First sentence
#people watch campusx
#people watch
#watch campusx
#people watch  -> column 2
#watch campusx -> column 4
#column:        0 1 2 3 4 5
#               ↓ ↓ ↓ ↓ ↓ ↓
#vector:        0 0 1 0 1 0
#Means:
#campusx watch  = 0
#campusx write  = 0
#eople watch   = 1   ← present
#people write   = 0
#watch campusx  = 1   ← present
#write comment  = 0
print(bow_ngram[1].toarray())
print(bow_ngram[2].toarray())
```

```python
# @title Data Preprocessing

from nltk import sent_tokenize
from gensim.utils import simple_preprocess
import nltk
nltk.download('punkt_tab')
```

```python
story = []
for filename in os.listdir('data'):
  if filename == '.ipynb_checkpoints':
    pass
  f = open(os.path.join('data', filename))
  corpus = f.read()
  #sent_tokenize() breaks a large paragraph into individual sentences.
  # This is called Sentence Tokenization.
  raw_sent = sent_tokenize(corpus)
  #print(raw_sent)
  for sent in raw_sent:
    #Clean and tokenize the sentence
    # Means : Converts text to lowercase
    #Removes punctuation
    #Splits the sentence into individual words
    story.append(simple_preprocess(sent))

```

```python
story
```

```python
story[0]
```

```python
# Initialize gensim model
model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)
```

```python
# Vector Representation
# Step 1 : Create an empty brain 🧠
# Step 2 : Build vocabulary
  # Read all words
  # Find unique words
  # Assign internal IDs/numbers
# Step 3: Learn relationships
  # i.e king ↔ queen
        # cat  ↔ dog
        # king ↔ powerful
        # cat  ↔ milk
# Final Conversion :
  #king  → [0.21, -0.55, 0.81, ...]
  #queen → [0.19, -0.51, 0.79, ...]
  #cat   → [-0.72, 0.14, 0.33, ...]
model.build_vocab(story)
```

```python
# Train Model
#An epoch means:
  #One complete round of learning through the entire dataset.
#Corpus : How many training examples/sentences are there?
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
# Get all the vectors
vec = model.wv.get_normed_vectors()
vec
```

```python
#y[0] → "king"
#y[1] → "queen"
#y[2] → "man"
#y[3] → "woman"
y = model.wv.index_to_key
len(y)

y
```

```python

from sklearn.decomposition import PCA

pca = PCA(n_components=3)
#X[0] → [2.1, 1.5]   → king
#X[1] → [2.0, 1.7]   → queen
#X[2] → [1.4, 0.8]   → man
#X[3] → [1.3, 1.0]   → woman
X = pca.fit_transform(vec)
X
#Relationship
#y[0]  ←→  X[0]
#y[1]  ←→  X[1]
#y[2]  ←→  X[2]
#y[3]  ←→  X[3]
```
